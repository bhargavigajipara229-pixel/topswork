```python
Q1 = df["Temperature_C"].quantile(0.25)
Q3 = df["Temperature_C"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["Temperature_C"] < lower) |
    (df["Temperature_C"] > upper)
]

print(outliers)
```

```python
Q1 = df["IceCream_Sales"].quantile(0.25)
Q3 = df["IceCream_Sales"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Lower Bound:", lower)
print("Upper Bound:", upper)

outliers = df[
    (df["IceCream_Sales"] < lower) |
    (df["IceCream_Sales"] > upper)
]

print("Outliers:")
print(outliers)
print("Number of outliers:", len(outliers))
```

```python
df_clean = df[
    (df["IceCream_Sales"] >= lower) &
    (df["IceCream_Sales"] <= upper)
].copy()

print("Before:", len(df))
print("After:", len(df_clean))
```

```python
X= df[["Temperature_C"]]
y=df["IceCream_Sales"]
```

```python
X_train,X_test,y_train,y_test=train_test_split(

X,
y,
test_size=0.2,
random_state=12
)
```
